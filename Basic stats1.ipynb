```python
#Load the dataset into a data analysis tool or programming environment
import numpy as np
import pandas as pd
df = pd.read_csv("C:\\Users\\vadde\\OneDrive\\Desktop\\sales_data_with_discounts.csv")
df
```

```python
#Identify numerical columns in the dataset.
df.select_dtypes(include="number").columns
```

```python
#Calculate the mean, median, mode, and standard deviation for these columns.
df["Volume"]
df["Avg Price"]
df["Total Sales Value"]
df["Discount Rate (%)"]
df["Discount Amount"]
df["Net Sales Value"]
```

```python
#df["Volume"]
print("Mean:",np.round(df["Volume"].mean(),2))
print("Medain:",np.round(df["Volume"].median(),2))
print("Mode:",np.round(df["Volume"].mode(),2))
print("Standard Deviation:",np.round(df["Volume"].std(),2))

#df["Avg Price"]
print("Mean:",np.round(df["Avg Price"].mean(),2))
print("Medain:",np.round(df["Avg Price"].median(),2))
print("Mode:",np.round(df["Avg Price"].mode(),2))
print("Standard Deviation:",np.round(df["Avg Price"].std(),2))

#df["Total Sales Value"]
print("Mean:",np.round(df["Total Sales Value"].mean(),2))
print("Medain:",np.round(df["Total Sales Value"].median(),2))
print("Mode:",np.round(df["Total Sales Value"].mode(),2))
print("Standard Deviation:",np.round(df["Total Sales Value"].std(),2))

#df["Discount Rate (%)"]
print("Mean:",np.round(df["Discount Rate (%)"].mean(),2))
print("Medain:",np.round(df["Discount Rate (%)"].median(),2))
print("Mode:",np.round(df["Discount Rate (%)"].mode(),2))
print("Standard Deviation:",np.round(df["Discount Rate (%)"].std(),2))

#df["Discount Amount"]
print("Mean:",np.round(df["Discount Amount"].mean(),2))
print("Medain:",np.round(df["Discount Amount"].median(),2))
print("Mode:",np.round(df["Discount Amount"].mode(),2))
print("Standard Deviation:",np.round(df["Discount Amount"].std(),2))

#df["Net Sales Value"]
print("Mean:",np.round(df["Net Sales Value"].mean(),2))
print("Medain:",np.round(df["Net Sales Value"].median(),2))
print("Mode:",np.round(df["Net Sales Value"].mode(),2))
print("Standard Deviation:",np.round(df["Net Sales Value"].std(),2))
```

```python
#Provide a brief interpretation of these statistics.
#Here the statistical measures for each numerical column can be seen above
'''
Mean - The average value of the data.

Median - The middle value of the data.

Mode - The value that comes most often.

Standard Dviation - Shows how much the values spread out from the average.
'''
```

```python
#relationship of numerical and categorical variables in the dataset.
#This is just for one realtionship between one categorical and one numerical variable
import seaborn as sns
plt.figure(figsize=(12,6))
sns.boxplot(x="Date",y="Volume", data=df)
plt.title("Distribution of model by Date")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45,ha='right')
plt.tight_layout()
plt.show()
```

```python
##Like that we have to make a relationship with every categorical and numberical variables so we have to use loops
#To visualize the distribution and relationship of numerical and categorical variables in the dataset.
import seaborn as sns
numerical_cols = df.select_dtypes(include="number").columns
categorical_cols = df.select_dtypes(include="object").columns

for num_col in numerical_cols:
  for cat_col in categorical_cols:
    plt.figure(figsize=(12,6))
    sns.boxplot(x=cat_col,y=num_col,data=df)
    plt.title(f"Distribution of {num_col} by {cat_col}")
    plt.xlabel(cat_col)
    plt.ylabel(num_col)
    plt.xticks(rotation=45, ha = "right")
    plt.tight_layout
    plt.show()
```

```python
#Analyze the distribution (e.g., skewness, presence of outliers) and provide inferences
#for numerical columns
import matplotlib.pyplot as plt
import seaborn as sns

# Numerical columns -[histogram]
df.hist(figsize=(10,8), bins=30)
plt.show()

# Categorical columns - [countplot]
for col in df.select_dtypes(include=["object"]).columns:
    sns.countplot(x=df[col])
    plt.title(f"Count of {col}")
    plt.xticks(rotation=45)
    plt.show()

```

```python
#	Boxplots:
#Create boxplots for numerical variables to identify outliers and the interquartile range.
#Discuss any findings, such as extreme values or unusual distributions.

import seaborn as sns
numeric_cols = df.select_dtypes(include=np.number).columns
for num_col in numerical_cols:
  plt.figure(figsize=(12,6))
  plt.title(f'Box Plot for {num_col}')
  sns.boxplot(df[num_col])
  plt.xticks(rotation=45,ha="right")
  plt.tight_layout()
  plt.show()
```

```python
#In summary, the box plots reveal the presence of outliers in most of the numerical columns, particularly in 'Volume', 'Avg Price', 'Total Sales Value', 'Discount Amount', and 'Net Sales Value'.
#These outliers represent instances of unusually high values
#The distributions of 'Volume', 'Avg Price', 'Total Sales Value', 'Discount Amount', and 'Net Sales Value' are skewed to the right.
#The 'Discount Rate (%)' has a more symmetric distribution but with some low outliers.
```

```python
#Bar Chart Analysis for Categorical Column:
#Identify categorical columns in the dataset.
#Create bar charts to visualize the frequency or count of each category.
#Analyze the distribution of categories and provide insights.
import seaborn as sns
import matplotlib.pyplot as plt

categoric_col = df.select_dtypes(include="object").columns

for cat_col in categoric_col:
    plt.figure(figsize=(12, 6))
    df[cat_col].value_counts().plot(kind='bar')
    plt.title(f'Bar chart for {cat_col}')
    plt.xlabel("cat_col")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
#Standardization of Numerical Variables
#Objective: To scale numerical variables for uniformity, improving the dataset’s suitability for analytical models.
numeric_col=df.select_dtypes(include=np.number).columns
from sklearn.preprocessing import StandardScaler
SS=StandardScaler()
SS_X=SS.fit_transform(df[numeric_col])
SS_X=pd.DataFrame(SS_X)
SS_X.columns=list(numeric_col)
SS_X
```

```python
#Centering: Before standardization, each numerical column had its own mean and spread.
#After standardization, all numerical columns will have a mean of approximately 0.
#This centers the distributions around zero.
#Scaling: Before standardization, the numerical columns had different scales and standard deviations.
#After standardization, all numerical columns will have a standard deviation of approximately 1
#This scales the distributions so that they have similar ranges relative to their means.
```
